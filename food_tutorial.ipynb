```python
import json
from food import FoodItem
```

```python

with open("products.json", "r", encoding="utf-8") as f:
    data = json.load(f)

products = []
def walk(node):
    if isinstance(node, dict):
        if "products" in node:
            products.extend(node["products"])
        for v in node.get("subcategories", {}).values():
            walk(v)
walk(data["home"])

items = [FoodItem.from_dict(p) for p in products]
```

```python
print(len(items))
item = items[0]
print(item.name)                              # "Coffee Creamer, French Vanilla"
print(item.nutrients.calories)                # Quantity(value=35.0, unit='cal')
print(item.nutrients.macronutrients.sodium)   # Quantity(value=10.0, unit='mg')
print(item.nutrients.macronutrients.fat.total)  # Quantity(value=1.5, unit='g')
```
