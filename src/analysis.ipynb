```python
from models.food import parse_foods
from scripts.rank_foods import (
    nutrient_per_dollar,
    sort_foods_by_nutrient_price,
    resolve_nutrient,
    inspect_nutrient_ranking
)

foods = parse_foods()

by_protein = sort_foods_by_nutrient_price(foods, "protein")
by_fiber = sort_foods_by_nutrient_price(foods, "fiber")
by_calories = sort_foods_by_nutrient_price(foods, "calories")
by_iron = sort_foods_by_nutrient_price(foods, "iron")
by_vitamin_a = sort_foods_by_nutrient_price(foods, "vitamin a")
by_fat = sort_foods_by_nutrient_price(foods, "fat.total")
by_trans = sort_foods_by_nutrient_price(foods, "fat.+ trans")
```

```python
nutrient = "vitamin a"
ranked = sort_foods_by_nutrient_price(foods, nutrient)

for food in ranked:
    quantity = resolve_nutrient(food, nutrient)
    score = nutrient_per_dollar(food, nutrient)
    # if score < 10:
    print(f"{food.name:<55} {score:>8.2f} {quantity.unit or ''}/$")
```
